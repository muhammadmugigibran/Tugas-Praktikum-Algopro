```python
# Inisialisasi variabel awal
hari = 1
saldo = 50000

# Perulangan selama hari <= 30
while hari <= 30:
    # Minta input pengeluaran dari user
    pengeluaran = int(input(f"Pengeluaran hari ke-{hari}: "))
    
    # Kurangi saldo dengan pengeluaran
    saldo -= pengeluaran
    
    # Cek jika saldo habis (kurang dari 0)
    if saldo < 0:
        print(f"Saldo habis di hari ke-{hari}. Doraemon menghentikan pencatatan!")
        break
    
    # Jika saldo masih cukup (else dari kondisi saldo < 0)
    else:
        print(f"Sisa saldo hari ke-{hari}: Rp{saldo}")
        hari += 1
```

```python
# Buat list pengeluaran mingguan Nobita
pengeluaran_mingguan = [20000, 0, 15000, 0]

# Buat perulangan dengan for i in range(4)
for i in range(4):
    # Akses nilai pengeluaran berdasarkan indeks i
    pengeluaran = pengeluaran_mingguan[i]
    minggu_ke = i + 1
    
    # Gunakan if jika pengeluaran_mingguan == 0, maka continue
    if pengeluaran == 0:
        print(f"Minggu ke-{minggu_ke}: Nobita tidak jajan, dilewati analisisnya.")
        continue
        
    # Jika tidak nol, tampilkan pengeluaran minggu itu
    print(f"Minggu ke-{minggu_ke}: Total pengeluaran Rp{pengeluaran}")

# Tambahkan else setelah perulangan
else:
    print("Analisis mingguan selesai.")
```

```python
# Gunakan for untuk perulangan outer bulan (1-3)
for bulan in range(1, 4):
    print(f"\nBulan ke-{bulan}")
    
    total_bulanan = 0
    terjadi_break = False
    
    # Gunakan for untuk perulangan inner minggu (1-4)
    for minggu in range(1, 5):
        # Langkah 3: Input pengeluaran tiap minggu
        pengeluaran = int(input(f"  Pengeluaran minggu ke-{minggu}: "))
        
        # Gunakan kondisi jika pengeluaran > 20000, maka break
        if pengeluaran > 20000:
            print("  Pengeluaran terlalu besar! Pencatatan bulan ini dihentikan.")
            terjadi_break = True
            break
            
        # Tambahkan total bulanan jika pengeluaran aman
        total_bulanan += pengeluaran
    
    # Tambahkan else setelah perulangan inner untuk mencetak total bulanan
    # Note: else pada for hanya berjalan jika loop inner TIDAK terkena break
    else:
        print(f"Total pengeluaran bulan ke-{bulan}: Rp{total_bulanan}")

# Setelah semua bulan selesai, tampilkan ringkasan
print("\n--- Analisis Laporan Bulanan Selesai ---")
```

```python
# Inisiasi saldo awal
saldo_awal = 300000
saldo = saldo_awal

# Gunakan for untuk perulangan bulan (1-12)
for bulan in range(1, 13):
    print(f"\n=== Bulan ke-{bulan} ===")
    
    # Gunakan while untuk perulangan minggu (<= 4)
    minggu = 1
    while minggu <= 4:
        # Buatkan inputan apakah minggu tersebut libur (y/n)?
        libur = input(f"Apakah minggu ke-{minggu} libur? (y/n): ").lower()
        
        # Jika y: lanjutkan ke minggu berikutnya (continue)
        if libur == 'y':
            print("Minggu ini dilewati (continue).")
            minggu += 1  # Increment tetap dilakukan agar tidak terjadi infinite loop
            continue
            
        # Jika n: minta input pengeluaran dan kurangi saldo
        elif libur == 'n':
            pengeluaran = int(input("Masukkan pengeluaran minggu ini: "))
            saldo -= pengeluaran
            
            # Jika saldo < 0: break dari perulangan
            if saldo < 0:
                print("Saldo habis! Simulasi langsung berhenti.")
                break
        
        # Lanjutkan ke minggu berikutnya jika tidak libur dan saldo aman
        minggu += 1
        
    # Periksa apakah loop while berhenti karena saldo habis (break)
    if saldo < 0:
        break
        
    # Gunakan else setelah while untuk mencetak saldo akhir bulan
    else:
        print(f"Saldo tersisa akhir bulan {bulan}: Rp{saldo}")

# Gunakan else setelah for untuk mengucapkan keberhasilan
else:
    print(f"\nNobita berhasil menabung Rp{saldo} dari total Rp{saldo_awal}!")
```
