```python
# 1. Variabel Data Diri
nama = "Giant"
usia = 20
berat_badan = 97.5          # float
tinggi_badan = 1.75         # float
status_olahraga = False     # boolean

# 2. Kasus konversi catatan data berat badan bertipe string -> float
berat_catatan_string = "97.5"
berat_catatan_float = float(berat_catatan_string)

# Kalimat pengantar menggunakan f-string (menggunakan berat_badan awal)
kalimat_pengantar = f"Halo, Saya {nama}. Usia saya {usia} tahun. Berat badan saya {berat_badan} kg."

# Output Center
lebar = 60

# Pembatas
print("-" * lebar)

# Output kalimat pengantar
print(kalimat_pengantar.center(lebar))

# Output tambahan untuk verifikasi hasil konversi string ke float
print(f"Catatan berat badan terbaru (setelah konversi): {berat_catatan_float} kg".center(lebar))
print(f"Tipe data hasil konversi: {type(berat_catatan_float)}".center(lebar))

# Pembatas
print("-" * lebar)
```

```python
# Perhitungan BMI (Body Mass Index)

# 1. Data Giant
berat_badan =97.5 # kg
tinggi_badan = 1.75 # m

# 2. Perhitungan BMI
bmi = berat_badan / (tinggi_badan ** 2)
bmi_pembulatan = round(bmi, 2)

# Output Center
lebar = 60

# 3. Output hasil perhitungan BMI
print("-" * lebar)
print("Hasil Perhitungan BMI".center(lebar))
print(f"BMI Asli: {bmi}".center(lebar))
print(f"BMI Setelah Pembulatan: {bmi_pembulatan}".center(lebar))
print("-" * lebar)
```

```python
# Target Minum Air Harian

target_minum = 2000 # mL
per_botol = 350 # mL

# Perhitungan Pembagian
hasil_pembagian_asli = target_minum / per_botol # True division
jumlah_botol = target_minum // per_botol # Floor division
sisa_air = target_minum % per_botol # Modulo operation

# Centered
lebar = 60

# Output hasil perhitungan target minum air harian
print("=" * lebar)
print(f"Hasil Perhitungan Target Minum Air Harian".center(lebar))
print("-" * lebar)
print(f"Hasil Asli (True Division): {hasil_pembagian_asli}".center(lebar))
print(f"Jumlah botol (Floor Division): {jumlah_botol}".center(lebar))
print(f"Sisa air (Modulo Operation): {sisa_air}".center(lebar))
print("=" * lebar)
```

```python
# Pembulatan Kebutuhan Kalori
kalori = 2100.5 # kalori

import math

kalori_pembulatan_ke_atas = math.ceil(kalori) # Pembulatan ke atas
kalori_pembulatan_ke_bawah = math.floor(kalori) # Pembulatan ke bawah

# Centered
lebar = 60

print("=" * lebar)
print(f"Kalori Asli: {kalori}".center(lebar))
print(f"Kalori Setelah Pembulatan ke Atas: {kalori_pembulatan_ke_atas}".center(lebar))
print(f"Kalori Setelah Pembulatan ke Bawah: {kalori_pembulatan_ke_bawah}".center(lebar))
print("=" * lebar)
```
