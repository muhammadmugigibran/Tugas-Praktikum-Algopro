```python
# TOTAL HARGA BELANJAAN NOBITA

# List dengan harga barang belanjaan Nobita
list_barang = ["Apel", "Susu","Roti",]


# List dengan jumlah barang belanjaan Nobita dengan Tuple
harga_jumlah = [5000, 3, 12000, 2, 8000, 1]

# Menghitung total harga belanjaan Nobita (manual)
total_apel = harga_jumlah[0] * harga_jumlah[1]     # 5000 * 3
total_susu = harga_jumlah[2] * harga_jumlah[3]     # 12000 * 2
total_roti = harga_jumlah[4] * harga_jumlah[5]     # 8000 * 1

# Harga masing-masing barang belanjaan Nobita
print(f"Harga Apel: {total_apel}")
print(f"Harga Susu: {total_susu}")
print(f"Harga Roti: {total_roti}")

# Total harga belanjaan Nobita
total_harga = total_apel + total_susu + total_roti
print(f"Total harga belanjaan Nobita: {total_harga}")

```

```python
# PEMBUATAN STRUK BELANJAAN NOBITA

# Data Pembeli
nama_pembeli = "Nobita Nobi"
nama_toko = "Toko Sembako Doraemon"
list_barang = ["Apel", "Susu","Roti",]
harga_jumlah = [5000, 3, 12000, 2, 8000, 1]

total_apel = harga_jumlah[0] * harga_jumlah[1]     # 5000 * 3
total_susu = harga_jumlah[2] * harga_jumlah[3]     # 12000 * 2
total_roti = harga_jumlah[4] * harga_jumlah[5]     # 8000 * 1
total_harga = total_apel + total_susu + total_roti

# Cetak Struk Belanja menggunakan f-string dan escape sequences
GREEN = '\033[92m'
RESET = '\033[0m'
print(f"{GREEN}========================================")
print(f"\t{nama_toko.upper()}")
print("========================================")
print(f"Nama Pembeli\t: {nama_pembeli}")
print("----------------------------------------")
print("Barang\t\tHarga\t  Qty\tTotal")
print("----------------------------------------")
print(f"{list_barang[0]}\t\tRp{harga_jumlah[0]}\t   {harga_jumlah[1]}\tRp{total_apel}")
print(f"{list_barang[1]}\t\tRp{harga_jumlah[2]}\t   {harga_jumlah[3]}\tRp{total_susu}")
print(f"{list_barang[2]}\t\tRp{harga_jumlah[4]}\t   {harga_jumlah[5]}\tRp{total_roti}")
print("----------------------------------------")
print(f"TOTAL BELANJA\t\t\t Rp{total_harga}")
print("========================================")
print("     Terima Kasih Telah Berbelanja!")
print("========================================")
print(RESET, end="")  # Reset warna teks ke default
```
