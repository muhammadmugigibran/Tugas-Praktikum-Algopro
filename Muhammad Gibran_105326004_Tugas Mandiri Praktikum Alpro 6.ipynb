PROGRAM KEDISPLINAN NOBITA

```python
# Variabel Global
skor_nobita = 0


# Fungsi untuk memperbarui skor
def update_skor(nilai):
    global skor_nobita
    skor_nobita += nilai


# Fungsi belajar
def belajar(jam_belajar):
    if jam_belajar < 1:
        return 50
    elif jam_belajar < 3:
        return 25
    else:
        return 10


# Fungsi tidur
def tidur(jam_tidur):
    if jam_tidur < 1:
        return -10
    elif jam_tidur <= 2:
        return 20
    else:
        return -20


# Fungsi main bersama Doraemon
def main_bersama_doraemon(jumlah_gadget, waktu_main):
    if waktu_main <= 1 and jumlah_gadget <= 2:
        return 15
    elif waktu_main <= 2 and jumlah_gadget <= 3:
        return 5
    else:
        return -30


# Fungsi menyontek
def menyontek():
    return -50
```

```python
# Fungsi pengendali aktivitas (mengembalikan True untuk lanjut, False untuk keluar)
def kontrol():
    print("\n=== PROGRAM PENINGKAT DISIPLIN NOBITA ===")
    print("1. Belajar")
    print("2. Tidur")
    print("3. Main bersama Doraemon")
    print("4. Menyontek")
    print("5. Keluar & Lihat Evaluasi Akhir")  # <-- Menambahkan menu keluar
    print()
    
    pilihan = input("\nPilih aktivitas (1-5): ")

    if pilihan == "1":
        jam = float(input("Berapa jam Nobita belajar hari ini? "))
        skor = belajar(jam)
    elif pilihan == "2":
        jam = float(input("Berapa jam Nobita tidur siang hari ini? "))
        skor = tidur(jam)
    elif pilihan == "3":
        gadget = int(input("Berapa gadget yang digunakan Nobita? "))
        waktu = float(input("Berapa jam Nobita main gadget? "))
        skor = main_bersama_doraemon(gadget, waktu)
    elif pilihan == "4":
        skor = menyontek()
    elif pilihan == "5":
        return False  # <-- Berhenti perulangan jika pilih 5
    else:
        print("Pilihan tidak valid, silakan pilih menu 1-5")
        return True

    update_skor(skor)
    print("Skor aktivitas :", skor)
    print("Total skor Nobita saat ini :", skor_nobita)
    return True
```

```python
# Fungsi evaluasi akhir
def evaluasi(skor):
    if skor >= 80:
        return "Siswa Teladan"
    elif skor >= 50:
        return "Cukup Rajin"
    else:
        return "Pemalas"
```

```python
# ==================================
# PROGRAM UTAMA (DENGAN PERULANGAN)
# ==================================
lanjut = True
while lanjut:
    lanjut = kontrol()  # Akan terus berulang selama fungsi kontrol mengembalikan True

print("\n=== HASIL EVALUASI AKHIR ===")
print("Skor Akhir :", skor_nobita)
print("Predikat   :", evaluasi(skor_nobita))
```

```python
# ==================================
# PROGRAM PENINGKAT DISIPLIN NOBITA
# ==================================

# Variabel Global
skor_nobita = 0


# Fungsi untuk memperbarui skor
def update_skor(nilai):
    global skor_nobita
    skor_nobita += nilai


# Fungsi belajar
def belajar(jam_belajar):
    if 1 <= jam_belajar <= 3 :
        return 15
    elif 4 <= jam_belajar <= 6:
        return 25
    else:
        return 50


# Fungsi tidur
def tidur(jam_tidur):
    if 1 <= jam_tidur < 6:
        return -10
    elif 6 <= jam_tidur <= 8:
        return 20
    else:
        return -20


# Fungsi main bersama Doraemon
def main_bersama_doraemon(jumlah_gadget, waktu_main):
    if 1 <= waktu_main < 3 and jumlah_gadget > 1:
        return 15
    elif 3 <= waktu_main <= 5 and jumlah_gadget > 4:
        return 5
    else:
        return -30


# Fungsi menyontek
def menyontek():
    return -50


# Fungsi pengendali aktivitas (mengembalikan True untuk lanjut, False untuk keluar)
def kontrol():
    print("\n=== PROGRAM PENINGKAT DISIPLIN NOBITA ===")
    print("1. Belajar")
    print("2. Tidur")
    print("3. Main bersama Doraemon")
    print("4. Menyontek")
    print("5. Keluar & Lihat Evaluasi Akhir")  # <-- Menambahkan menu keluar
    print()
    
    pilihan = input("\nPilih aktivitas (1-5): ")

    if pilihan == "1":
        jam = float(input("Berapa jam Nobita belajar hari ini? "))
        skor = belajar(jam)
    elif pilihan == "2":
        jam = float(input("Berapa jam Nobita tidur siang hari ini? "))
        skor = tidur(jam)
    elif pilihan == "3":
        gadget = int(input("Berapa gadget yang digunakan Nobita? "))
        waktu = float(input("Berapa jam Nobita main gadget? "))
        skor = main_bersama_doraemon(gadget, waktu)
    elif pilihan == "4":
        skor = menyontek()
    elif pilihan == "5":
        return False  # <-- Berhenti perulangan jika pilih 5
    else:
        print("Pilihan tidak valid, silakan pilih menu 1-5")
        return True

    update_skor(skor)
    print("Skor aktivitas :", skor)
    print("Total skor Nobita saat ini :", skor_nobita)
    return True


# Fungsi evaluasi akhir
def evaluasi(skor):
    if skor >= 80:
        return "Siswa Teladan"
    elif skor >= 50:
        return "Cukup Rajin"
    else:
        return "Pemalas"


# ==================================
# PROGRAM UTAMA (DENGAN PERULANGAN)
# ==================================
lanjut = True
while lanjut:
    lanjut = kontrol()  # Akan terus berulang selama fungsi kontrol mengembalikan True

print("\n=== HASIL EVALUASI AKHIR ===")
print("Skor Akhir :", skor_nobita)
print("Predikat   :", evaluasi(skor_nobita))
```
