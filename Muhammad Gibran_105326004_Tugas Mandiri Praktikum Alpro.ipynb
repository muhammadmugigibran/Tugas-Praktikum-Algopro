```python
# PERENCANAAN NAMA TOKO DORAYAKI
# Meminta variabel dari orang lain
nama_makanan = input("Ketik 1 nama makanan")
nama_karakter = input("Ketik 1 nama karakter")
kata_sifat = input("Ketik 1 kata sifat")

# Gabungan dari ketiga variabel
nama_toko = f"{nama_makanan} {nama_karakter} {kata_sifat}"

# Hasil rekomendasi toko dorayaki
print("\n--- Rekomendasi Nama Toko Dorayaki ---")
print(f"Nama toko yang diusulkan adalah: {nama_toko}")

# Hitungan Modal Awal Toko Dorayaki
# Doraemon = d, Nobita = n, Shizuka = s
d = 555000
n = 333000
s = 222000
total_modal = d + n + s

#Total Modal
print("--- Total Modal Toko Dorayaki ---")
print(f"Total modal yang terkumpul: Rp{total_modal:,}")
print()

# Hasil Penjualan 1 Hari
jumlah_dorayaki = 350
harga_dorayaki = 5000

# Total Pendapatan
total_pendapatan = jumlah_dorayaki * harga_dorayaki
print("--- Total Pendapatan Toko Dorayaki ---")
print(f"Total pendapatan: Rp{total_pendapatan:,}")
print()

# Keuntungan dari modal dan penjualan
keuntungan = total_pendapatan - total_modal

print("--- Laporan Keuangan Toko Dorayaki ---")
print(f"Total Pendapatan: Rp{total_pendapatan:,}")
print(f"Total Modal: Rp{total_modal:,}")
print(f"Total Keuntungan: Rp{keuntungan:,}")
```
