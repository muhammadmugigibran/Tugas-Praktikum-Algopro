```python
raw_data = """
baling_baling_bambu|transportasi|low|12
pintu_ke_mana_saja|ruang-waktu|high|5

mesin_waktu|ruang-waktu|extreme|1
  senter_pengecil|manipulasi|medium|8
selimut_ajaib|pertahanan|medium|7

konyaku_penerjemah|komunikasi|low|20
cetak_pintu|manipulasi|medium|10
"""

# 1. Bersihin string & cek data awal
baris_bersih = []
for baris in raw_data.splitlines():
    baris = baris.strip()
    if baris != "":
        baris_bersih.append(baris)

teks_bersih = "\n".join(baris_bersih)
jumlah_alat = len(baris_bersih)
ada_waktu = "waktu" in teks_bersih.lower()
preview = teks_bersih[:30]

print("Jumlah alat:", jumlah_alat)
print("Ada kata 'waktu':", ada_waktu)
print("Preview:", preview)
print()

# 2. Pecah string jadi nested list
data_alat = []
for baris in baris_bersih:
    bagian = baris.split("|")
    nama = bagian[0].strip()
    kategori = bagian[1].strip()
    bahaya = bagian[2].strip()
    stok = int(bagian[3].strip())
    data_alat.append([nama, kategori, bahaya, stok])

print("Nested list:")
for item in data_alat:
    print(item)
print()

# 3. Pisahkan alat berbahaya dan tidak berbahaya
berbahaya = []
tidak_berbahaya = []

for item in data_alat:
    if item[2] == "extreme" or item[2] == "high":
        berbahaya.append(item[0])
    elif item[2] == "low":
        tidak_berbahaya.append(item[0])

print("Alat berbahaya:", berbahaya)
print("Alat tidak berbahaya:", tidak_berbahaya)
print()

# 4. Urutkan nama alat pakai sorted()
urut_stok = [x[0] for x in sorted(data_alat, key=lambda x: x[3])]

level_bahaya = {"low": 1, "medium": 2, "high": 3, "extreme": 4}
urut_bahaya = [x[0] for x in sorted(data_alat, key=lambda x: level_bahaya[x[2]])]

urut_panjang = [x[0] for x in sorted(data_alat, key=lambda x: len(x[0]))]

print("Urut stok:", urut_stok)
print("Urut bahaya:", urut_bahaya)
print("Urut panjang nama:", urut_panjang)
print()

# 5. Update data
data_alat.append(["pistol_pembalik_waktu", "ruang-waktu", "extreme", 2])
data_alat = [item for item in data_alat if item[0] != "cetak_pintu"]

for item in data_alat:
    if item[0] == "senter_pengecil":
        item[3] += 5

print("Setelah update:")
for item in data_alat:
    print(item)
print()

# 6. Cetak laporan
print("=" * 40)
print("Laporan Alat Doraemon".center(40))
print("=" * 40)

for i, item in enumerate(data_alat):
    nama_tampil = item[0].replace("_", " ").title()
    print(f"Nama: {nama_tampil}")
    print(f"Kategori: {item[1]}")
    print(f"Tingkat Bahaya: {item[2]}")
    print(f"Stok tersedia: {item[3]}")
    if i < len(data_alat) - 1:
        print("-" * 40)
```
