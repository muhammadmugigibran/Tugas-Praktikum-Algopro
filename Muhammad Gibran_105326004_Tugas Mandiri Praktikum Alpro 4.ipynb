```python
print("=== SMART VEHICLE INSURANCE ANALYZER (DORABIKE) ===")

# 1. Variabel Input dari Pengguna
usia_pengemudi = int(input("Masukkan usia pengemudi: "))
jenis_kendaraan = input("Masukkan jenis kendaraan (mobil/motor): ").lower()
nilai_kendaraan = float(input("Masukkan nilai kendaraan (dalam Rupiah): "))
area_penggunaan = input("Masukkan area penggunaan (perkotaan/pedesaan): ").lower()
sistem_pengaman = input("Apakah kendaraan memiliki sistem pengaman? (ya/tidak): ").lower()

# 2. Menentukan Kategori Risiko Awal berdasarkan Jenis Kendaraan (if-else)
if jenis_kendaraan == "motor":
    risiko = "sedang"
else:
    risiko = "rendah"

# 3. Penyesuaian Risiko berdasarkan Usia Pengemudi (operator boolean & if)
if usia_pengemudi < 20 or usia_pengemudi > 60:
    risiko = "tinggi"

# 4. Penyesuaian Risiko berdasarkan Area Penggunaan (operator boolean & if)
if area_penggunaan == "perkotaan" and risiko == "rendah":
    risiko = "sedang"
elif area_penggunaan == "perkotaan" and risiko == "sedang":
    risiko = "tinggi"

# 5. Penyesuaian Risiko berdasarkan Sistem Pengaman (if-elif-else)
if sistem_pengaman == "tidak":
    if risiko == "rendah":
        risiko = "sedang"
    elif risiko == "sedang":
        risiko = "tinggi"
    # Jika sudah tinggi, tetap tinggi

# 6. Menghitung Premi Asuransi berdasarkan Kategori Risiko (if-elif-else)
if risiko == "tinggi":
    persentase_premi = 0.1
elif risiko == "sedang":
    persentase_premi = 0.07
else:
    persentase_premi = 0.05

premi_asuransi = persentase_premi * nilai_kendaraan

# 7. Menentukan Pesan Doraemon menggunakan Conditional Expression
pesan = "Doraemon menyarankan asuransi tambahan untuk perlindungan penuh!" if risiko == "tinggi" else "Risiko kendaraan masih aman menurut Doraemon"

# 8. Output Hasil Analisis
print("\n--- HASIL ANALISIS ASURANSI DORABIKE ---")
print("1. Kategori Risiko:", risiko.capitalize())
print("2. Premi Asuransi yang Perlu Dibayar: Rp", f"{premi_asuransi:,.2f}")
print("3. Pesan Doraemon:", pesan)
```
